```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
y = x**2
y
```

```python
plt.plot(x,y)
plt.title("Matplotlib")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)

plt.subplot(2,2,2)
plt.plot(x**2,x**2)

plt.subplot(2,2,3)
plt.plot(x,x)


plt.subplot(2,2,4)
plt.plot(y,x)
```

```python
fig = plt.figure(figsize=[12,8],dpi=100)
axis1 = fig.add_axes([0.2,0.2,0.4,0.4])
axis1.plot(x,x)
axis2 = fig.add_axes([0.1,0.1,0.4,0.4])
axis2.plot(x,x)
```

```python
plt.figure(figsize=[12,8])
```

```python
plt.scatter(x,y)
```

```python
plt.hist(y)
```

```python
plt.boxplot(x)
```

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig('basicplot.png')
```
